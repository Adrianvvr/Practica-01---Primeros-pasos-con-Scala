## Ejercicio 4 — Función para determinar si una mano se pasa de 21

```scala
def bust(puntuacion:Int):Boolean={
    if (puntuacion > 21)
        true
    else
        false
}
bust(18)
bust(21)
bust(22)
bust(30)
```

* **Función `Bust`:** Determina si una puntuación supera los 21 puntos.
* **Lógica:** Recibe un número entero (`Int`). Si es mayor a 21, devuelve `true`; si es 21 o menos, devuelve `false`.
* **Pruebas:** Las últimas líneas llaman a la función con los valores 18, 21, 22 y 30 para verificar el resultado en cada caso.

## Ejercicio 5 — Comparación de dos manos
Devuelve la mano mayor y si son iguales, 0

```scala
def maxHand(valor1:Int, valor2:Int):Int={
    if (valor1 == valor2)
        0
    else if (valor1 > valor2)
        valor1
    else
        valor2
}
println(maxHand(17, 19))
println(maxHand(20, 18))
println(maxHand(21, 21))
```

## Ejercicio 6 — Decidir el ganador de una partida

```scala
def ganador(handA:Int, handB:Int):Int={
    if(bust(handA) && bust(handB))
        0
    else if(bust(handA) && !bust(handB))
        handB
    else if(!bust(handA) && bust(handB))
        handA
    else
        maxHand(handA, handB)
}
println(ganador(26, 20))
println(ganador(18, 22))
println(ganador(24, 25))
println(ganador(17, 19))
println(ganador(21, 20))
```

## Ejercicio 7 — Arrays y mutabilidad

```scala
val jugadores = Array("Alex", "Chen", "Marta")
println(jugadores.mkString(", "))
jugadores(0) = "Sindhu"
println(jugadores.mkString(", "))
jugadores(1) = 500
```

* **Mutabilidad interna:** Aunque la variable `jugadores` se declara con `val` (referencia inmutable), la estructura `Array` en Scala es mutable por naturaleza. No puedes asignar un nuevo array a la variable, pero sí cambiar los datos que contiene.
* **Tipado estricto:** Al inicializar el array con textos, Scala infiere automáticamente que es un `Array[String]`. Por lo tanto, no permite asignar un valor numérico como `500` (tipo `Int`), lanzando un error de incompatibilidad (*type mismatch*).
* **Reasignar vs. Modificar:** Reasignar implica intentar que la variable apunte a un objeto completamente distinto (prohibido al usar `val`). Modificar un elemento solo altera el contenido interno del mismo objeto (permitido por la estructura del `Array`).

## Ejercicio 9 — Recorrer un Array con while

```scala
val manos = Array(17, 24, 21, 19, 26)
var i = 0
while(i < manos.length){
    println(manos(i))
    println(bust(manos(i)))
    i+= 1
}
```

* **Contador mutable:** La variable `i` debe ser `var` porque su valor necesita actualizarse (aumentar en 1) en cada iteración para poder avanzar a la siguiente posición del array.
* **Falta de incremento:** Si no se incrementa `i`, su valor siempre sería 0. La condición del `while` siempre se cumpliría, provocando un **bucle infinito** que bloquearía la ejecución.
* **Estilo imperativo:** Se evidencia en el control manual del flujo y el cambio de estado: el uso de variables mutables (`var i`), la instrucción explícita de cómo iterar paso a paso (`while`), y la modificación directa del estado en cada ciclo (`i += 1`).

## Ejercicio 10 — Listas e inmutabilidad

```scala
val jugadores = List("Alex", "Chen", "Marta")
println(jugadores)
val nuevaLista = "Sindhu" :: jugadores
println(jugadores)
println(nuevaLista)
println(jugadores.length)
println(jugadores.reverse)
```

* **Inmutabilidad:** En Scala, la estructura `List` es inmutable por defecto. Una vez creada, sus elementos no pueden ser modificados, añadidos ni eliminados.
* **Operador `::` (cons):** Como la lista original (`jugadores`) está protegida contra cambios, al usar `::` para añadir un elemento al principio, Scala crea obligatoriamente una **nueva lista** (`nuevaLista`). Esta nueva lista enlaza el nuevo elemento con la lista original, la cual se mantiene intacta.

## Ejercicio 12 — Operadores relacionales y lógicos

```scala
val handA = 18
val handB = 21
val handC = 25

handA > handB

handB == 21

handC != 21

handA <= 21 && handB <= 21

handA > 21 || handC > 21

!(handB > 21)
```

| Pregunta / Descripción | Expresión en Scala | Resultado |
| :--- | :--- | :--- |

## Ejercicio 13 — foreach y funciones como valores

```scala
val manos = Array(17, 24, 21, 26, 18)
var i = 0

while (i < manos.length) {
    println(bust(manos(i)))
    i += 1
}
println()

manos.foreach(mano => println(bust(mano)))
```

* **Necesita un contador:** La primera versión (el bucle `while`), que utiliza la variable `i` para llevar la cuenta de la posición actual del array.
* **Necesita una variable `var`:** También la primera versión, ya que requiere declarar `var i = 0` para poder incrementar y actualizar el índice en cada iteración.
* **Estilo funcional:** La segunda versión (`manos.foreach(...)`). Este enfoque es más declarativo, evita mutar estados (no usa `var` ni contadores explícitos) y trata a las funciones como valores, pasándole la lógica directamente a la colección para que ella se encargue de iterar.
